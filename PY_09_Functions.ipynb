#### Positional Agrument
    The position defined in function must be passed in same order.

```python
def details(name,age,city):
    print(f"Name is {name}")
    print(f"Age is {age}")
    print(f"City is {city}")
details('TOM',25,'New York')
details(25,'New York','TOM')
```

#### Default Argument
    If the value from the user is not defined in the function it will keep the default one given.

```python
def network(username, password, port = 8080):
    print(username, password, port)
network("xyz@gmail.com","123456789")   # Port no will taken by default
network("xyz@gmail.com","123456789", 440)  # It will take user port no
```

#### Keyword Argument
    Regardless of the position of the variable we can pass in any order by specifying the variable

```python
def details(name,age,city):
    print(f"Name is {name}")
    print(f"Age is {age}")
    print(f"City is {city}")
details('SAM',30,'CA')
details(30,'CA','SAM')
details(age = 30,city = 'CA',name = 'SAM')
```

#### Variable Length Arguments
    In cases we don't know the length of arguments to be passed we can use variable length argument
    Syntax 
        def function_name(*args):
            code

```python
def printing_value(*args):
    print(type(args))
    for i in args:
        print(i)
printing_value(10,60,56,64642,3415,3513)
print('')
printing_value(10,60,56,64642,3415,3513,34,65446,654,8797421,16615,16526)
```

#### Variable length Keyword Arguments

```python
def temp(**kwargs):
    print(kwargs)
temp(name='Sam', age=30, city='CA', Dept='IT', Sal= 2546535263)
```

```python
def temp1(**kwargs):
    print(kwargs)
    for i,j in kwargs.items():
        print(i,j)
temp1(name='Sam',age=30,city='CA',Dept='IT',Sal= 2546535263)
```

### Write a python function which take username and password validate them

```python
def validate_user(username, password):
    valid_username = 'abc@gamil.com'
    valid_pass = 'Pass@123'
    return username == valid_username and password == valid_pass
print(validate_user('abc@gmail.com','Pass@123'))
print(validate_user('abc@gamil.com','Pass@123'))
```

### Write a python code to calculate the grade for respective percentage

```python
def calulate_grade(percentage):
    if percentage >= 90:
        return "A+"
    elif percentage >= 80:
        return "B+"
    elif percentage >= 70:
        return "C+"
    else:
        return "D"
print(calulate_grade(50))
print(calulate_grade(80))
```

## Types of Variable
    1) Global Variable -- Defined the outside function
    2) Local Variable -- Define Inside the Function
    Local variable can also be assigned as global by using the global command inside a function. As the function is called the local variable will automatically become global variable
        Syntax
            def func()
                global var
                var = 10

```python
var_name = 'Python' 
def show():
    print(var_name)
show()
```

```python
var_name = 'Python'   # Global Variable 
def show():
    var_name = 'Inside Function'   # Local Variable
    print(var_name)
show()
```

```python
var_name = 'Python'   # Global Variable 
def show():
    var_name = 'Inside Function'   # Local Variable
    print(var_name)
print(var_name)
```

```python
a = 10  # global variable
def show_a():
    print(a)
def show_b():
    b = 20  # local variable
    print(b)
show_a()
show_b()
print(a)
print(b)
```

```python
b = 1000
def abc():
    c= 2000   #local variable can be use in same function you cannot call it outside
    print(b)
def xyz():
    print(b)
    print(c)
xyz()
```

```python
b = 1000
def abc():
    b=10
    print(b)
def xyz():
    print(b)
abc()    # 10
xyz()    # 1000
print(b) # 1000
```

```python
b = 1000
def abc():
    global b
    b=10  
    print(b)
def xyz():
    b=15
    print(b)
abc()    # 10     # Local b becomes Global b and value changes from 1000 to 10
xyz()    # 15     # b value is changed from 10 to 15, this 15 is only for xyz()
print(b) # 10     # this print(b) is outside any func, so the latest globally assigned b value is assigned here 
```

```python
x = 100
def cool():
    global x
    x = 200
    print(x)
cool()
print(x)
```

## Inner Function and Outer Function

```python
a = 10
def show():
    print(a)
show()
```

```python
def a():    #Outer Function
    print("Hello from Outer Function")    
    def b(): #Inner Function
        print("Hello from Inner Function")
    print("Bye")
a()
```

```python
def a():    #Outer Function
    print("Hello from Outer Function")    
    def b(): #Inner Function
        print("Hello from Inner Function")
    b()
    print("Bye")
a()
```

```python
def a():                                  #Outer Function
    print("Hello from Outer Function")   
    b()                                   # Cannot call a function before defining - UnboundLocalError
    def b():                              #Inner Function
        print("Hello from Inner Function")    
    print("Bye")
a()
```

```python
def m1():                   # Outer Function
    a = 100  
    print(a)
    def m2():               # Inner Function
        a = 200
        print(a)
        def m3():           # inner inner function
            print(a)        # Func m2 and m3 will get called automatically when func m1 is called
        m3()                # func m3 cannot be called seperately outside func m2
    m2()                    # similiarly m2 cannot be called outside func m1
```

```python
m1()
```

```python
m2()
```

```python
def m1():                 # Outer Function
    a = 100  
    print(a)
    def m2():             # Inner Function
        a = 200
        print(a)
        def m3():         # inner inner function
            print(a)
    m3()                  # Cannot call m3 outside m2
    m2()
```

```python
m1()
```
